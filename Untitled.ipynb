```python
import requests
import json

def fetch_pakistan_railway_data():
    overpass_url = "https://overpass-api.de/api/interpreter"
    
    query = """
    [out:json][timeout:300];
    area["ISO3166-1"="PK"][admin_level=2]->.pk;
    (
      node["railway"="station"](area.pk);
      way["railway"="station"](area.pk);
      relation["railway"="station"](area.pk);
    );
    out center;
    (
      way["railway"="rail"](area.pk);
    );
    out geom;
    """
    
    print("Fetching Pakistan Railway topology from Overpass API...")
    response = requests.post(overpass_url, data={'data': query})
    
    if response.status_code == 200:
        data = response.json()
        with open("network.json", "w", encoding="utf-8") as f:
            json.dump(data, f, indent=4)
        print(f"Successfully saved {len(data.get('elements', []))} network elements to network.json.")
    else:
        print(f"Error {response.status_code}: {response.text}")

if __name__ == "__main__":
    fetch_pakistan_railway_data()
```

```python
import requests, time, json

ENDPOINTS = [
    "https://overpass-api.de/api/interpreter",
    "https://overpass.kumi.systems/api/interpreter",
    "https://maps.mail.ru/osm/tools/overpass/api/interpreter",
]

HEADERS = {"User-Agent": "ThermoRail-hackathon/0.1 (contact: your-email@example.com)"}

QUERY = """
[out:json][timeout:180];
area["ISO3166-1"="PK"][admin_level=2]->.pk;
way["railway"="rail"](area.pk);
out geom;
"""

def fetch_pakistan_railway_data(retries=3):
    for attempt in range(retries):
        for url in ENDPOINTS:
            try:
                print(f"Trying {url} (attempt {attempt+1})...")
                r = requests.post(url, data={"data": QUERY},
                                  headers=HEADERS, timeout=(10, 240))
                if r.status_code == 200:
                    return r.json()
                print(f"  HTTP {r.status_code}: {r.text[:200]}")
            except requests.exceptions.RequestException as e:
                print(f"  Failed: {type(e).__name__}")
        time.sleep(10 * (attempt + 1))  # back off before the next round
    raise RuntimeError("All Overpass endpoints failed")

data = fetch_pakistan_railway_data()
with open("pk_railways.json", "w") as f:
    json.dump(data, f)
print(len(data["elements"]), "ways saved")
```

```python
from math import radians, sin, cos, asin, sqrt

def way_km(e):
    g = e["geometry"]
    total = 0
    for a, b in zip(g, g[1:]):
        la1, lo1, la2, lo2 = map(radians, (a["lat"], a["lon"], b["lat"], b["lon"]))
        h = sin((la2-la1)/2)**2 + cos(la1)*cos(la2)*sin((lo2-lo1)/2)**2
        total += 2 * 6371 * asin(sqrt(h))
    return total

def group(e):
    t = e.get("tags", {})
    u, s = t.get("usage", "missing"), t.get("service")
    if u in ("main", "branch") and s is None: return "running"
    if u == "missing" and s is None:          return "ambiguous"
    return "other"

from collections import defaultdict
km = defaultdict(float); n = defaultdict(int)
for e in els:
    g = group(e); km[g] += way_km(e); n[g] += 1
for g in km: print(g, n[g], "ways", round(km[g]), "km")
```

```python
amb = [e for e in els if group(e) == "ambiguous"]
print(sum(1 for e in amb if "name" in e["tags"]), "have a name")
print(Counter(e["tags"].get("gauge", "none") for e in amb))
print(Counter(e["tags"].get("electrified", "none") for e in amb))
```

```python
running_nodes = set()
for e in els:
    if group(e) == "running":
        running_nodes.update(e["nodes"])

def touches_running(e):
    return any(n in running_nodes for n in e["nodes"])

from collections import defaultdict
stats = defaultdict(lambda: [0, 0.0])
for e in amb:
    key = ("gauge" if "gauge" in e["tags"] else "no gauge",
           "touches running" if touches_running(e) else "isolated")
    stats[key][0] += 1
    stats[key][1] += way_km(e)

for k, (n, km_) in sorted(stats.items()):
    print(k, n, "ways", round(km_), "km")
```

```python
parent = {}
def find(x):
    parent.setdefault(x, x)
    while parent[x] != x:
        parent[x] = parent[parent[x]]
        x = parent[x]
    return x
def union(a, b):
    parent[find(a)] = find(b)

use = [e for e in els if group(e) in ("running", "ambiguous")]
for e in use:
    for n in e["nodes"][1:]:
        union(e["nodes"][0], n)

comp_km, comp_ways = defaultdict(float), defaultdict(list)
for e in use:
    r = find(e["nodes"][0])
    comp_km[r] += way_km(e)
    comp_ways[r].append(e)

top = sorted(comp_km.items(), key=lambda x: -x[1])
print(len(top), "components")
for r, km_ in top[:10]:
    print(round(km_), "km,", len(comp_ways[r]), "ways")
```

```python
giant = top[0][0]
amb_stats = defaultdict(lambda: [0, 0.0])
for e in amb:
    where = "giant" if find(e["nodes"][0]) == giant else "other component"
    kind = "gauge" if "gauge" in e["tags"] else "no gauge"
    amb_stats[(kind, where)][0] += 1
    amb_stats[(kind, where)][1] += way_km(e)

for k, (n, km_) in sorted(amb_stats.items()):
    print(k, n, "ways", round(km_), "km")
```

```python
def confidence(e):
    g = group(e)
    if g == "running":
        return "high"
    if g == "ambiguous":
        in_giant = find(e["nodes"][0]) == giant
        return "medium" if in_giant else "low"
    return "exclude"

conf = Counter(confidence(e) for e in els)
print(conf)
```

```python
import geopandas as gpd
from shapely.geometry import LineString

rows = []
for e in els:
    c = confidence(e)
    if c == "exclude":
        continue
    t = e.get("tags", {})
    rows.append({
        "osm_id": e["id"],
        "name": t.get("name"),
        "usage": t.get("usage"),
        "gauge": t.get("gauge"),
        "electrified": t.get("electrified"),
        "confidence": c,
        "length_km": way_km(e),
        "geometry": LineString([(p["lon"], p["lat"]) for p in e["geometry"]]),
    })

gdf = gpd.GeoDataFrame(rows, crs="EPSG:4326")
gdf.to_file("pk_rail_segments.geojson", driver="GeoJSON")
print(len(gdf)); print(gdf.confidence.value_counts())
```

```python
from shapely.ops import substring

PIECE_KM = 1.0   # change this once you know your temperature resolution

g = gdf.to_crs(32642)   # UTM 42N, so lengths are in meters
keep = ["osm_id", "name", "usage", "gauge", "electrified", "confidence"]

out = []
for row in g.itertuples():
    L = row.geometry.length
    n = max(1, round(L / (PIECE_KM * 1000)))
    step = L / n
    for i in range(n):
        out.append({
            **{k: getattr(row, k) for k in keep},
            "piece": i,
            "geometry": substring(row.geometry, i * step, (i + 1) * step),
        })

pieces = gpd.GeoDataFrame(out, crs=32642)
pieces["length_km"] = pieces.length / 1000
pieces = pieces.to_crs(4326)
pieces.to_file("pk_rail_pieces.geojson", driver="GeoJSON")

print(len(pieces))
print(pieces.length_km.describe())
```

```python
import requests, time
import pandas as pd

URL = "https://archive-api.open-meteo.com/v1/archive"
START, END = "2026-05-01", "2026-06-30"
HOURLY = "temperature_2m,soil_temperature_0cm"
BATCH = 25

cells = pieces[["cell_id", "cell_lat", "cell_lon"]].drop_duplicates().reset_index(drop=True)

def fetch_batch(b):
    params = {
        "latitude": ",".join(f"{x:.2f}" for x in b.cell_lat),
        "longitude": ",".join(f"{x:.2f}" for x in b.cell_lon),
        "start_date": START, "end_date": END,
        "hourly": HOURLY,
    }
    for attempt in range(4):
        r = requests.get(URL, params=params, timeout=120)
        if r.status_code == 200:
            js = r.json()
            return js if isinstance(js, list) else [js]
        print("  HTTP", r.status_code, r.text[:150])
        time.sleep(30 * (attempt + 1))
    raise RuntimeError("batch failed")

def summarize(cells_subset, out_file):
    rows = []
    for i in range(0, len(cells_subset), BATCH):
        b = cells_subset.iloc[i:i + BATCH]
        res = fetch_batch(b)
        for (_, c), js in zip(b.iterrows(), res):
            h = pd.DataFrame(js["hourly"])
            rows.append({
                "cell_id": c.cell_id,
                "api_lat": js["latitude"], "api_lon": js["longitude"],
                "air_max_c": h.temperature_2m.max(),
                "air_p95_c": h.temperature_2m.quantile(0.95),
                "surf_max_c": h.soil_temperature_0cm.max(),
                "surf_p95_c": h.soil_temperature_0cm.quantile(0.95),
            })
        pd.DataFrame(rows).to_csv(out_file, index=False)   # saved after every batch
        print(i + len(b), "/", len(cells_subset))
        time.sleep(5)
    return pd.DataFrame(rows)

test = summarize(cells.head(3), "cell_heat_test.csv")
print(test)
```

```python
import pandas as pd

heat = pd.read_csv("cell_heat.csv")
heat["soil_minus_air_p95"] = heat.soil_p95_c - heat.air_p95_c
heat["heat_index_c"] = heat[["air_p95_c", "soil_p95_c"]].max(axis=1)

# tier cutoffs as quantiles of the 787 cells; edit to taste
q = heat.heat_index_c.quantile([0.5, 0.8, 0.95])
def tier(v):
    if v >= q[0.95]: return "extreme"
    if v >= q[0.8]:  return "high"
    if v >= q[0.5]:  return "elevated"
    return "moderate"
heat["risk_tier"] = heat.heat_index_c.apply(tier)
print(q)
print(heat.risk_tier.value_counts())

out = pieces.merge(heat.drop(columns=["api_lat", "api_lon"]), on="cell_id", how="left")
assert len(out) == len(pieces) and out.heat_index_c.notna().all()
out.to_file("pk_rail_heat.geojson", driver="GeoJSON")

# eyeball check
colors = {"moderate": "#4575b4", "elevated": "#fee090", "high": "#f46d43", "extreme": "#a50026"}
out.plot(color=out.risk_tier.map(colors), linewidth=1.2, figsize=(10, 12))

cols = ["cell_id", "api_lat", "api_lon", "air_p95_c", "soil_p95_c", "soil_minus_air_p95"]
print(heat.nlargest(8, "soil_minus_air_p95")[cols])
print(heat.nsmallest(8, "heat_index_c")[cols])
```

```python
import pandas as pd

heat = pd.read_csv("cell_heat.csv")
print("cells:", len(heat),
      "| distinct model cells:", heat[["api_lat", "api_lon"]].drop_duplicates().shape[0])

def tiers(s):
    q = s.quantile([0.5, 0.8, 0.95])
    return pd.cut(s, [-1, q[0.5], q[0.8], q[0.95], 99],
                  labels=["moderate", "elevated", "high", "extreme"], right=False)

heat["tier_air"]  = tiers(heat.air_p95_c)
heat["tier_soil"] = tiers(heat.soil_p95_c)
heat["tier_both"] = tiers(heat[["air_p95_c", "soil_p95_c"]].max(axis=1))

print(pd.crosstab(heat.tier_air, heat.tier_both))
```

```python
heat["rank_air"]   = heat.air_p95_c.rank(pct=True)
heat["rank_soil"]  = heat.soil_p95_c.rank(pct=True)
heat["heat_score"] = (heat.rank_air + heat.rank_soil) / 2
heat["tier_rank"]  = tiers(heat.heat_score)

print(pd.crosstab(heat.tier_air,  heat.tier_rank))
print(pd.crosstab(heat.tier_soil, heat.tier_rank))

new = ["cell_id", "heat_score", "tier_air", "tier_soil", "tier_rank"]
g2 = g.rename(columns={"risk_tier": "tier_max"}).merge(heat[new], on="cell_id", how="left")
assert len(g2) == len(g) and g2.heat_score.notna().all()
for c in ["tier_air", "tier_soil", "tier_rank"]:
    g2[c] = g2[c].astype(str)
g2.to_file("pk_rail_heat_v2.geojson", driver="GeoJSON")
```
